# Power Method



Run from the repository folder or any folder below it. Requires NumPy, SciPy and Numba.


```python
import time
import sys
from pathlib import Path
import numpy as np
from scipy.sparse import csr_matrix, issparse

ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "common.py").is_file()), None)
if ROOT is None:
    raise FileNotFoundError("Start the notebook inside Numerical_Linear_Algebra_Algorithms")
sys.path.insert(0, str(ROOT))
from common import loader, data_path, lu_solve, relative_residual
from Eigenvalue_Methods.power_method import power_method


def timed(fn, *args):
    start = time.perf_counter()
    result = fn(*args)
    return result, time.perf_counter() - start
```

## 3×3


```python
A = np.array([[4., 1., 0.], [1., 3., 1.], [0., 1., 2.]])
x_true = np.ones(3)
b = A @ x_true
rng = np.random.default_rng(42)
Z = rng.normal(size=(A.shape[1], 3))
_ = power_method(A)
result, seconds = timed(power_method, A)
print("Seconds:", seconds)
lam, x, info = result
print("Eigenvalue:", lam)
print(info)
reference = np.linalg.eigvalsh(A)
print("Small reference eigenvalues:", reference)
assert info["converged"]
assert abs(lam - float(reference[-1])) < 1e-6
```
